```python
import math

x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))
x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))

d = math.sqrt(math.pow(x2 - x1, 2) + math.pow(y2 - y1, 2))

print(f"The distance between these two points are: {d}")

#1. The math library helped simplify my work because I didn't have to make commands from scratch.
#2. Because of the library, the functions "math.sqrt" and "math.pow" were easy to use.
#3. The program would be very long, complex, and time consuming without the sqrt() and pow(),
```
