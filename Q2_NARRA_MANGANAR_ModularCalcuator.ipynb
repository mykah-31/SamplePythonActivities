```python
operation = str(input("Enter operation to be used: "))
solving = float(input("Enter number #1: "))
solving_2 = float(input("Enter number #2: "))

#make the functions
def add(a, b):
  addition = a + b
  return addition

def sub(a, b):
  subtraction = a - b
  return subtraction

def mul(a, b):
  multiplication = a * b
  return multiplication

def div(a, b):
  division = a / b
  return division

#Execute the program
if operation == "addition" or "Addition":
  print(add(solving, solving_2))

elif operation == "subtraction" or "Subtraction":
  print(sub(solving, solving_2))

elif operation == "multiplication" or "Multiplication":
  print(mul(solving, solving_2))

elif operation == "Division" or "division":
  print(div(solving, solving_2))

else:
  print("Sorry, this operation is not available.")

```
